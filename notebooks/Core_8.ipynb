## 1- Transformación de datos.

### Revision de Data

Los datos se encuentra limpios, y pueden ser tratos.

```python
import pandas as pd

# Carga dat
df = pd.read_csv('../data/retail_sales_dataset.csv')
df.head (10) #mostrar las primeras 10 filas del data frame

#Exploración
df_tail = df.tail(5) 
print("-"*100)
print(df_tail) # muestra las primeras 5 filas

print("-"*100)
df_info = df.info()
print(df_info) #muestra la información de la tabla

print("-"*100)
df_describe = df.describe()
print(df_describe) #muestra los descriptivos de la tabla

df_types = df.dtypes
print("-"*100)
print(df_types) #muestra conteo de datos

df_genero = df['Product Category'].value_counts()
print("-"*100)
print (df_genero) #muestra conteo valores por categoria productos

df_unicos = df['Gender'].unique() #No existe la variable tienda por lo que se reemplaza por Gender
print("-"*100)
print (df_unicos) #muestra valores únicos de Gender
```

### Creación de columnas

Creación de la columna Pocentaje de ventas

```python

# Calcular el porcentaje de ventas totales
df['Percentage_TA'] = df['Total Amount'].transform(lambda x: x / x.sum() * 100).round(2)

print(df)
```

### Clasificación de datos

Se clasifican los valores de ventas: Alta>1400; Baja<100, 1000<=Medio<=1400. 

```python
# Clasificar los productos por rango de ventas (monetario)
df['Classification Total Amount'] = df['Total Amount'].apply(lambda x: 'Alta' if x > 1400 else ('Baja' if x < 1000 else 'Media'))
print(df)
```

### 2- Agrupación y Agregación.

Se realizan dos agrupaciones, primero por producto y luego por producto y año.

Para cada uno de estos grupos se determinan funciones de agregación.

```python
#Agruación por un columna y luego por dos columnas.
Gr_A= df.groupby('Product Category')
Gr_B = df.groupby(['Product Category','Age'])

#Aplicación de funciones de agregación

Grupo_ventas1=Gr_A[['Total Amount','Percentage_TA']].agg(['sum'])
Grupo_ventas_y_año=Gr_B[['Total Amount']].agg(['mean','min','max','std'])

print("-"*100)
print(Grupo_ventas1)
print("="*100)
print("")


print("-"*100)
print(Grupo_ventas_y_año)
print("="*100)
print("")
```

## 3- Análisis Personalizado con apply

Se agrega la columna 'Cantidad_normalizada', la cual incorpora la normalización de la cantidad vendida.

```python
# Calcular la normalización de la columna cantidad deventas.

max_cantidad = df['Quantity'].max()
min_cantidad = df['Quantity'].min()
df['Cantidad_normalizada'] = df['Quantity'].apply(lambda x: (x - min_cantidad) / (max_cantidad - min_cantidad))

print(df)
```
