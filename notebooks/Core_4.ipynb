### Carga y procesamiento de los datos

```python

import pandas as pd

# Carga data
df = pd.read_csv('retail_sales_dataset.csv')

# 1. Inspección de la data
print("ESTRUCTURA E INTEGRIDAD")
df.info()  # muestra la información de las columnas y memoria usada

# Ver n de los nulos
print("\nConteo de nulos por columna:")
print(df.isnull().sum()) #muestra las columnas y tipo

# 2. Limpieza por poner porque no tiene nulo
# Usa solo filas con datos
df = df.dropna() 
df['Date'] = pd.to_datetime(df['Date'])
```

### Exploracion de los datos

```python


#  Total de ventas y promedio por categoría
exploracion = df.groupby('Product Category')['Total Amount'].agg(['sum', 'mean'])
exploracion.columns = ['Total Ventas', 'Promedio Ventas']

print("Análisis por Categoría")
print(exploracion)

# 2. Identificar categorías con mayores y menores ventas
cat_mayor = exploracion['Total Ventas'].idxmax()
cat_menor = exploracion['Total Ventas'].idxmin()

print(f"\nCategoría con MAYOR éxito: {cat_mayor} (${exploracion.loc[cat_mayor, 'Total Ventas']:.2f})")
print(f"Categoría con MENOR éxito: {cat_menor} (${exploracion.loc[cat_menor, 'Total Ventas']:.2f})")
```

### Manipulación de datos

```python

# 1. Filtrar solo las ventas de 'Electronics' 
categoria_target = 'Electronics'
df_filtrado = df[df['Product Category'] == categoria_target].copy()

# 2. Operaciones aritméticas adicionales
df_filtrado['Tax_19%'] = df_filtrado['Total Amount'] * 0.19          # IVA 0.19 (multplicación)
df_filtrado['Descuento_Fijo'] = df_filtrado['Total Amount'] - 5      # descuento del 5% Resta
df_filtrado['Precio_Final'] = df_filtrado['Total Amount'] + 2        # se le suma el costo de la bolsa
df_filtrado['Venta_por_Unidad'] = df_filtrado['Total Amount'] / df_filtrado['Quantity'] #  por unidad (división)

print(f"\n--- Estadísticas adicionales para {categoria_target} ---")
print(df_filtrado[['Total Amount', 'Tax_19%', 'Venta_por_Unidad']].head())
```
